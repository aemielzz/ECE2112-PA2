### 1. NORMALIZATION PROBLEM

```python
import numpy as np 
```

```python
x = np.random.random((5,5))
```

```python
print (x)
```

```python
a = x.mean()
```

```python
b = x.std()
```

```python
normal = (x-a)/b
```

```python
print (normal)
```

```python
np.save ('X_normalized.npy', normal)
```

## 2. Divisible by 3 Problem

```python
a = np.arange(1,101)
e = np.square (a)
m = e.reshape (10,10)

print(m)
```

```python
c = m [m%5==0]
z = c.reshape(4,5)
print (z)
```
